```python
import numpy as np
from scipy import stats

sales = np.array([45, 52, 48, 55, 60, 58, 62, 65, 70, 68, 72, 150])

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Mode:", stats.mode(sales, keepdims=True).mode[0])
print("Range:", np.max(sales) - np.min(sales))
print("Variance:", np.var(sales, ddof=1))
print("Standard Deviation:", np.std(sales, ddof=1))

q1, q3 = np.percentile(sales, [25, 75])
print("IQR:", q3 - q1)
```
